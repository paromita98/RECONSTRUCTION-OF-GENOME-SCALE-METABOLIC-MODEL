```python
import cobra
from cobra import Reaction

# 1. Load model
model = cobra.io.read_sbml_model("E:/gapseq-output/Phenylobacterium-zucineum/fill_model/Phenylobacterium-zucineum.xml")
# 2. Mapping from media shorthand to actual SBML IDs
media_map = {
    "glc__D": "cpd00027_c0",
    "o2": "cpd00007_c0",
    "nh4": "cpd00013_c0",
    "pi": "cpd00009_c0",
    "h2o": "cpd00001_c0",
    "h": "cpd00067_c0",
    "na1": "cpd00971_c0",
    "k": "cpd00205_c0",
    "cl": "cpd00099_c0",
    "fe2": "cpd10515_c0",
    "fe3": "cpd10516_c0",
    "cobalt2": "cpd00149_c0",
    "mg2": "cpd00254_c0",
    "so4": "cpd00048_c0",
    "h2s": "cpd00239_c0",
    "zn2": "cpd00034_c0",
    "cu2": "cpd00058_c0",
    "mn2": "cpd00118_c0",
    "ca2": "cpd00063_c0",
    "fru__D": "cpd00082_c0"
}

# 3. Biomass coefficients
biomass_mets = {
    "glc__D": 5,
    "fru__D": 5
}
for k in media_map:
    if k not in biomass_mets:
        biomass_mets[k] = 100.0

# 4. Create biomass reaction
biomass_rxn = Reaction("BIOMASS_PSEUDO")
biomass_rxn.name = "Biomass pseudo reaction"
biomass_rxn.lower_bound = 0
biomass_rxn.upper_bound = 100

# 5. Add metabolites
for short_id, coeff in biomass_mets.items():
    met_id = media_map[short_id]
    try:
        met = model.metabolites.get_by_id(met_id)
        biomass_rxn.add_metabolites({met: coeff})
    except KeyError:
        print(f"⚠ Could not find metabolite {met_id} in model.")

# 6. Add biomass reaction to model and set as objective
model.add_reactions([biomass_rxn])
model.objective = "BIOMASS_PSEUDO"

# 7. Save updated model
cobra.io.write_sbml_model(model, "E:/gapseq-output/Phenylobacterium-zucineum/fill_model/Phenylobacterium-zucineum_with_biomass.xml")
print("✅ Biomass reaction added and saved to 'Minicystis-rosea.xml' ")

# Optional: test optimization
solution = model.optimize()
print("Growth rate:", solution.objective_value)
```
